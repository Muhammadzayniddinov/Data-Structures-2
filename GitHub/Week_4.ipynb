```python
# Challenge 1

# Time: O(n); space: O(n).
def is_balanced(s):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    for char in s:
        if char in '([{':
            stack.append(char)
        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return False
    return len(stack) == 0


assert is_balanced('(a[b]{c})') == True
assert is_balanced('([)]') == False
assert is_balanced('') == True
assert is_balanced(')') == False
assert is_balanced('((') == False
print('Challenge 1: tests passed')
```

```python
# Challenge 2

# Time: O(n) average; space: O(n).
def two_sum(nums, target):
    seen = {}
    for i, number in enumerate(nums):
        needed = target - number
        if needed in seen:
            return (seen[needed], i)
        seen[number] = i
    raise ValueError('No valid pair')


assert two_sum([2, 7, 11, 15], 9) == (0, 1)
assert two_sum([-3, 4, 3, 90], 0) == (0, 2)
assert two_sum([0, 4, 3, 0], 0) == (0, 3)
assert two_sum([3, 3], 6) == (0, 1)
print('Challenge 2: tests passed')
```

```python
# Challenge 3

class Node:
    def __init__(self, value, next_node=None):
        self.value = value
        self.next = next_node


# Time: O(n); extra space: O(1).
def reverse_list(head):
    previous = None
    current = head
    while current is not None:
        next_node = current.next
        current.next = previous
        previous = current
        current = next_node
    return previous


assert reverse_list(None) is None

single = Node(5)
assert reverse_list(single) is single
assert single.next is None

a, b, c = Node(1), Node(2), Node(3)
a.next = b
b.next = c
head = reverse_list(a)
assert head is c
assert c.next is b and b.next is a and a.next is None
print('Challenge 3: tests passed')
```
