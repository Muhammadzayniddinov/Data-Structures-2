```python
# Challenge 4

class StackNode:
    def __init__(self, value, next_node=None):
        self.value = value
        self.next = next_node


class Stack:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def push(self, value):
        self.head = StackNode(value, self.head)

    def pop(self):
        if self.is_empty():
            raise IndexError('Stack is empty')
        value = self.head.value
        self.head = self.head.next
        return value


# Time: enqueue O(1), dequeue O(1) amortized/O(n) worst case; space: O(n).
class QueueFromStacks:
    def __init__(self):
        self.in_stack = Stack()  # 'in' is a Python keyword.
        self.out_stack = Stack()

    def enqueue(self, x):
        self.in_stack.push(x)

    def dequeue(self):
        if self.out_stack.is_empty():
            while not self.in_stack.is_empty():
                self.out_stack.push(self.in_stack.pop())
        if self.out_stack.is_empty():
            raise IndexError('Queue is empty')
        return self.out_stack.pop()


q = QueueFromStacks()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
assert q.dequeue() == 1
assert q.dequeue() == 2
assert q.dequeue() == 3

q.enqueue(4)
q.enqueue(5)
assert q.dequeue() == 4
q.enqueue(6)
assert q.dequeue() == 5
assert q.dequeue() == 6

try:
    q.dequeue()
except IndexError:
    pass
else:
    raise AssertionError('Expected IndexError')
print('Challenge 4: tests passed')
```

```python
# Challenge 8

class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


# Time: O(n); extra space: O(h), where h is the tree height.
# Duplicate values are not allowed in a valid BST.
def is_valid_bst(root):
    def check(node, low, high):
        if node is None:
            return True
        if low is not None and node.value <= low:
            return False
        if high is not None and node.value >= high:
            return False
        return (check(node.left, low, node.value)
                and check(node.right, node.value, high))

    return check(root, None, None)


assert is_valid_bst(None) == True
assert is_valid_bst(Node(5)) == True
assert is_valid_bst(Node(5, Node(1), Node(8, Node(6), Node(9)))) == True
assert is_valid_bst(Node(5, Node(1), Node(8, Node(4), Node(9)))) == False
assert is_valid_bst(Node(5, Node(5), Node(8))) == False
print('Challenge 8: tests passed')
```
